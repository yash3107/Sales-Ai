```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Load the data from the Excel file
file_path = 'C:/Users/riju3/OneDrive/Desktop/Re-inforcement Learning1.xlsx'
data = pd.read_excel(file_path)

# Ensure the columns are properly named and contain "Sample Question" and "Sample Answer"
if 'Sample Question' not in data.columns or 'Sample Answer' not in data.columns:
    raise ValueError("The file must contain 'Sample Question' and 'Sample Answer' columns.")

# User inputs their question
user_question = input("Please enter your question: ")

# Calculate similarities between the user question and each sample question
similarities = []
for index, row in data.iterrows():
    sample_question = str(row['Sample Question'])

    # Vectorize the text with TF-IDF
    vectorizer = TfidfVectorizer()
    vectors = vectorizer.fit_transform([user_question, sample_question])

    # Calculate cosine similarity
    cosine_sim = cosine_similarity(vectors)[0][1]
    similarities.append(cosine_sim)

# Find the index of the most similar question
most_similar_index = similarities.index(max(similarities))

# Get the corresponding sample answer
most_similar_question = data.iloc[most_similar_index]['Sample Question']
most_similar_answer = data.iloc[most_similar_index]['Sample Answer']

# Display the results
print("\nMost Similar Question:")
print(most_similar_question)
print("\nSample Answer:")
print(most_similar_answer)
```
